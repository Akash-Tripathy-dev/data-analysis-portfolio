```python
import pandas as pd
import numpy as np

# Set Seed for reproducible messiness
np.random.seed(10)
total_records = 6000
```

```python
## Base pools with mixed casing , spaces and errors
```

```python
# Base pools with mixed casing, spaces, and errors
titles = ['Software Engineer', 'software engineer', 'SOFTWARE ENGINEER', 'Data Analyst', 'data analyst', 'Project Manager', '  Manager  ', 'UNKNOWN', None]
departments = ['Engineering', 'Finance', 'Human Resources', 'Marketing', 'N/A']

data = {
    'Employee_ID': [f"EMP-{np.random.randint(10000, 99999)}" for _ in range(total_records)],
    'Job_Title': np.random.choice(titles, size=total_records, p=[0.2, 0.2, 0.1, 0.15, 0.1, 0.1, 0.05, 0.05, 0.05]),
    'Department': np.random.choice(departments, size=total_records, p=[0.4, 0.2, 0.15, 0.2, 0.05]),
    'Years_Experience': np.random.choice([np.nan, -2, 1, 3, 5, 8, 12, 35, 99], size=total_records, p=[0.08, 0.02, 0.15, 0.2, 0.2, 0.15, 0.1, 0.08, 0.02]),
    'Performance_Tier': np.random.choice(['Level-1 (Low)', 'Level-2 (Medium)', 'Level-3 (High)', 'MISSING'], size=total_records, p=[0.25, 0.45, 0.25, 0.05]),
    'Annual_Salary_USD': np.random.uniform(25000, 185000, size=total_records)
}
df_raw = pd.DataFrame(data)

## Inject specific system errors: If Job title is missing , force Annual_Salary to NaN
df_raw.loc[df_raw['Job_Title'].isna(),'Annual_Salary_USD'] = np.nan
df_raw.to_csv('raw_employee_records.csv', index =False)
print("📥 Successfully compiled and created 'raw_employee_records.csv' (6,000 rows)!")
```
