```python
import pandas as pd
import numpy as np

# Set seed for reproducible banking records
np.random.seed(2026)
total_records = 7000

risk_scores = ['Class-A', 'Class-B', 'Class-C' ,'UNKNOWN', 'Class-B',None]
purposes = ['Debt Consolidation', 'Home Improvement', 'Small Business', 'Major Purchase', 'N/A']

data = {
    'Account_ID': [f"ACC-{np.random.randint(100000, 999999)}" for _ in range(total_records)],
    'Internal_Risk_Grade': np.random.choice(risk_scores, size=total_records, p=[0.40, 0.30, 0.15, 0.05, 0.05, 0.05]),
    'Loan_Purpose': np.random.choice(purposes, size=total_records, p=[0.45, 0.25, 0.15, 0.10, 0.05]),
    'Customer_Debt_To_Income_Ratio': np.random.choice([np.nan, -1.5, 12.5, 24.8, 38.2, 55.0, 99.9], 
                                                     size=total_records, p=[0.08, 0.02, 0.30, 0.25, 0.20, 0.13, 0.02]),
    'Loan_Default_Status': np.random.choice([0, 1], size=total_records, p=[0.88, 0.12]), 
    'Requested_Amount_Raw': [f" ${np.random.randint(5000, 45000):,}.00 " for _ in range(total_records)]
}

df_raw =pd.DataFrame(data)
df_raw.loc[df_raw['Internal_Risk_Grade'].isna(),'Customer_Debt_To_Income_Ratio']=np.nan

# Save to system storage 
df_raw.to_csv('raw_loan_portfolio.csv', index=False)
print("📥 'raw_loan_portfolio.csv' generated locally inside your folder!")
```
