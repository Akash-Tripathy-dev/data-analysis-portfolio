```python
import pandas as pd
import numpy as np

# Lock seed for structural consistency
np.random.seed(42)
total_logs = 7500

tower_regions = ['Rourkela_Sector-5','Bhubaneswar_DLF','Cuttack_LinkRoad','Sambalpur_Town']
network_types = ['5G','4G','3G']

data = {
    'Log_ID': [f"LOG-{np.random.randint(100000,999999)} " for _ in range(total_logs)],
    'Tower_Region':np.random.choice(tower_regions,size = total_logs,p=[0.4, 0.3, 0.15, 0.15]),
    'Network_Generation': np.random.choice(network_types, size=total_logs, p=[0.6, 0.35, 0.05]),
    
    # Download speed in Mbps (Injecting impossible negative values and massive glitched spikes)
    'Download_Speed_Mbps': np.random.choice([np.random.normal(150, 40), -50, 9999], size=total_logs, p=[0.97, 0.015, 0.015]),
    
    # Server error flags: True means the tower crashed during that hour
    'Server_Crash_Flag': np.random.choice([False, True], size=total_logs, p=[0.94, 0.06])
}

df_raw = pd.DataFrame(data)
df_raw.to_csv('raw_telecom_network_logs.csv', index=False)
print("📥 'raw_telecom_network_logs.csv' (7,500 rows) successfully saved to  notebook directory!")

```
