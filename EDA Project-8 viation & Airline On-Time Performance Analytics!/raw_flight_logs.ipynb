```python
import pandas as pd
import numpy as np

# Lock the seed for structural consistency
np.random.seed(77)
total_flights =5000

airlines = ['Indigo','Air India','Vistara','Akasa Air']
destinations = ['delhi_DEL',"Mumbai_BOM",'Bengaluru_BLR','Kolkata_CCU']

data = {
    'Flight_ID':[f"FLI-{np.random.randint(1000,9999)}" for _ in range(total_flights)],
    'Airline_Name': np.random.choice(airlines,size=total_flights,p=[0.45,0.25,0.20,0.10]),
    'Destinations_Hub' : np.random.choice(destinations,size=total_flights,p=[0.35,0.30,0.20,0.15]),
    
    #Departure Delay in minutes (Negative mean the flight left early)
    'Delay_Minutes': np.random.choice([np.random.normal(15,30), np.random.exponential(120),-5],size= total_flights, p=[0.70,0.25,0.05])
}

df_raw = pd.DataFrame(data)

# Round the minutes to whole numbers for realistic logging
df_raw['Delay_Minutes'] = df_raw['Delay_Minutes'].round().astype(int)

df_raw.to_csv('raw_flight_logs.csv', index=False)
print("📥 'raw_flight_logs.csv' (5,000 scheduled flights) saved to directory!")
```
