```python
import pandas as pd
import numpy as np

# Lock seed for consistency
np.random.seed(42)
total_records = 5500

food_categories = ['Vegetables', 'Dairy', 'Bakery', 'Snacks']
brands = ['Amul', 'Britannia', 'Fortune', 'Catch_Spices']
packaging_types = ['Packet', 'Bottle', 'Box', 'Loose_Poly']

data = {
    'Item_ID': [f"FOOD-{np.random.randint(1000, 9999)}" for _ in range(total_records)],
    'Food_Category': np.random.choice(food_categories, size=total_records, p=[0.35, 0.3, 0.25, 0.1]),
    
    # NEW EASY LOGIC: Clear, real-world descriptive columns
    'Item_Brand': np.random.choice(brands, size=total_records, p=[0.4, 0.3, 0.15, 0.15]),
    'Packaging_Type': np.random.choice(packaging_types, size=total_records, p=[0.4, 0.3, 0.2, 0.1]),
    
    'Weight_Grams': np.random.choice([np.nan, 100, 250, 500, 1000, 5000, -200, 99999], size=total_records, p=[0.05, 0.25, 0.3, 0.2, 0.15, 0.03, 0.01, 0.01]),
    'Item_Price_INR': np.random.uniform(10, 800, size=total_records),
    'Estimated_Calories': np.random.choice([np.nan, 50, 150, 350, 700], size=total_records, p=[0.25, 0.35, 0.3, 0.08, 0.02])
}
df_raw = pd.DataFrame(data)

# Data corruption simulation: if weight is negative, drop price to 0
df_raw.loc[df_raw['Weight_Grams'] <= 0, 'Item_Price_INR'] = 0.0

df_raw.to_csv('easy_grocery_inventory.csv', index=False)
print("📥 Success! 'easy_grocery_inventory.csv' generated with simple, everyday logic.")
```
