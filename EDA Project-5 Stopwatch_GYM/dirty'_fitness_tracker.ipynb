```python
import pandas as pd
import numpy as np

# Lock the seed for consitent progress tracking
np.random.seed(101)
total_days = 90

# GEnerate a continuous sequence of 90days 
date_range = pd.date_range(start='2026-03-01', periods=total_days, freq='D')
workout_types = ['Weightlifting', 'Cardio', 'Yoga', 'Rest_Day']

data = {
    # Messy text dates with shifting formats to practice standardization
    'Activity_Date': [d.strftime('%Y-%m-%d') if i % 5 != 0 else d.strftime('%d/%m/%Y') for i, d in enumerate(date_range)],
    'Workout_Type': np.random.choice(workout_types, size=total_days, p=[0.45, 0.35, 0.10, 0.10]),
    
    # Workout duration in minutes (with a few glitched negative values or empty slots)
    'Duration_Minutes': np.random.choice([np.nan, 45, 60, 75, 90, -15], size=total_days, p=[0.05, 0.3, 0.3, 0.2, 0.1, 0.05]),
    
    # Active calories burned
    'Calories_Burned': np.random.uniform(150, 750, size=total_days)
}

df_raw = pd.DataFrame(data)

# If it is a rest day , force duration and calories down to zero
df_raw.loc[df_raw['Workout_Type'] == 'Rest_Day',['Duration_Minutes', 'Calories_Burned']] = 0.0

df_raw.to_csv('dirty_fitness_tracker.csv', index= False)
print("📥 'dirty_fitness_tracker.csv' (90 days of gym logs) saved to your notebook directory!")
```
